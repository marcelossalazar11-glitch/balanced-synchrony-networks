# Balanced synchrony: multiple edge types

This notebook studies a six-vertex network with two edge types. **Each edge type must be preserved separately** when checking balance and computing graph automorphisms.

```python
import matplotlib.pyplot as plt

from balanced_synchrony import get_automorphisms, search_balanced_partitions
from balanced_synchrony.examples import two_edge_types_network
from balanced_synchrony.visualization import plot_partition
```

## Search and symmetry reduction

The graph has two adjacency matrices, named `black` and `blue`. The algorithm uses a multigraph isomorphism matcher with a **multiedge-aware** edge-type comparator.

```python
adjacency_by_type = two_edge_types_network()
automorphisms = get_automorphisms(adjacency_by_type)
result = search_balanced_partitions(adjacency_by_type)
print(f"Edge-type-preserving automorphisms: {len(automorphisms)}")
print(f"Partitions examined: {result.examined}")
print(f"Balanced representatives: {result.balanced_count}")
print(f"Exotic representatives: {result.exotic_count}")
assert (len(automorphisms), result.examined, result.balanced_count, result.exotic_count) == (8, 203, 5, 0)
```

```python
example = next(item for item in result.representatives if len(item.partition) > 1)
plot_partition(adjacency_by_type, example.partition, title="Two edge types — balanced synchrony")
plt.show()
```

## Interpretation

Balanced partitions describe candidate robust synchrony relations for coupled-cell network models. Classification as exotic refers to whether a partition arises from **network symmetry orbits**. This notebook does not calculate spectral stability or train a machine-learning model.
