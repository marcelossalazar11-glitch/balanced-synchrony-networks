# Balanced synchrony: one edge type

This notebook analyzes a 10-vertex network isomorphic to the **Petersen graph** and illustrates the distinction between balanced and exotic synchrony. See the [README](../README.md) for the mathematical definitions.

The adjacency convention is $A_{uv}=$ number of directed edges from $u$ to $v$. For an undirected edge, both entries are set to one.

```python
import matplotlib.pyplot as plt

from balanced_synchrony import analyze_partition, search_balanced_partitions
from balanced_synchrony.examples import frucht_network, petersen_network
from balanced_synchrony.visualization import plot_partition
```

## Exhaustive search on the Petersen graph

Partitions equivalent under graph automorphisms are counted once.

```python
adjacency = petersen_network()
result = search_balanced_partitions(adjacency)
print(f"Partitions examined: {result.examined:,}")
print(f"Balanced representatives: {result.balanced_count}")
print(f"Exotic representatives: {result.exotic_count}")
assert (result.examined, result.balanced_count, result.exotic_count) == (115975, 11, 0)
```

```python
example = next(item for item in result.representatives if len(item.partition) > 1)
plot_partition(adjacency, example.partition, title="Petersen graph — balanced synchrony")
plt.show()
```

## An exotic example: Frucht graph

The Frucht graph is 3-regular with a trivial automorphism group. Its one-block partition is balanced but cannot be induced by a subgroup of graph automorphisms. We classify only this selected partition instead of enumerating all partitions of twelve vertices.

```python
frucht = frucht_network()
frucht_partition = [list(range(12))]
frucht_result = analyze_partition(frucht, frucht_partition)
print(f"Exotic: {frucht_result.exotic}")
print(f"Blockwise stabilizer order: {frucht_result.stabilizer_order}")
print(f"Stabilizer orbit count: {len(frucht_result.stabilizer_orbits)}")
assert frucht_result.exotic
plot_partition(frucht, frucht_result.partition, title="Frucht graph — exotic synchrony")
plt.show()
```
